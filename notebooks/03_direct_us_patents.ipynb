```python
#Though it's US jurisdiction, for sanity, just double checking that language is english for all
langs = df["claims"].apply(
    lambda x: x[0].get("lang") if isinstance(x, list) and len(x) > 0 else None
)
langs.value_counts(dropna=False)
```

```python
# So the Claim column is a list but contains one dictionary, with keys of claims and language. 
# In that claims key, the value is another list of claim dictionaries where each has the key of claim_text and the value is the text
# This column is nested and odd looking.

# I'm just gonna extract ALL claim text into a per row, so flatten it:
def extract_all_claim_text(claim_obj):
    if not isinstance(claim_obj, list) or len(claim_obj) == 0: 
        return "" #For the 9 rows that don't have a list in their claims column (are float, NaN), just return an empty string
    
    claim_block = claim_obj[0]  # Take that first dictionary in [{ "claims": [...], "lang": "en" }]
    
    if "claims" not in claim_block: #just in case, confirm claims key exists
        return ""
    
    full_text = [] #initiate list
    
    for claim in claim_block["claims"]: #inside the list in the dictionary, [{[ ]}]
        text_parts = claim.get("claim_text", []) #grab the value of claim_text key. Return [] if key doesn't exist
        if isinstance(text_parts, list): #if text_parts we grabbed is a list (should be), then
            full_text.append(" ".join(text_parts)) #glue all items in list into one string, with space in between
    
    return " ".join(full_text) #combine all claims text
```

```python
df["claims_text"] = df["claims"].apply(extract_all_claim_text)
```

```python
#So title is in invention_title

#Grab the title using it:
empty_titles = empty_claims.apply(
    lambda row: row["biblio"].get("invention_title", [{}])[0].get("text")
    if isinstance(row["biblio"], dict)
    else None,
    axis=1
)
```

```python
#To search on google patents, also easier to get document number and not lens-id
def get_pub_number(biblio):
    if not isinstance(biblio, dict):
        return None
    
    pub = biblio.get("publication_reference", {})
    
    jurisdiction = pub.get("jurisdiction")
    doc_number = pub.get("doc_number")
    kind = pub.get("kind")
    
    if jurisdiction and doc_number and kind:
        return f"{jurisdiction}{doc_number}{kind}"

    return None


empty_pub_numbers = empty_claims["biblio"].apply(get_pub_number)

pd.set_option("display.max_colwidth", None) #resest with pd.reset_option("display.max_colwidth")

pd.DataFrame({
    "lens_id": empty_claims["lens_id"],
    "publication_number": empty_pub_numbers,
    "title": empty_titles
})
```

## Prepare files for API input
The next step is to prepare the text files to be inputted into API
- The three texts I can provide are Title, Abstract and Claims

<br>

<span style="background-color: yellow; color: black;">Note the limitations</span>
- If the claims references SEQ ID NO or Structures or Figures, I did not provide and include this information
- SEQ ID tables would significantly increase token length and are beyond the scope of this case study
- It would also require conversion of images into text, which would add preprocessing complexity

```python
def extract_title(biblio):
    if not isinstance(biblio, dict):
        return ""
    
    titles = biblio.get("invention_title", [])
    
    if isinstance(titles, list) and len(titles) > 0:
        return titles[0].get("text", "")
    
    return ""

df["title"] = df["biblio"].apply(extract_title)
```

```python
def extract_abstract(abstract_obj):
    if not isinstance(abstract_obj, list) or len(abstract_obj) == 0:
        return ""
    
    first_entry = abstract_obj[0]
    
    if isinstance(first_entry, dict):
        return first_entry.get("text", "")
    
    return ""

df["abstract_text"] = df["abstract"].apply(extract_abstract)
```

```python
#There are some rows where owner column is filled and is different from applicant
import ast

# 1. extract applicants
def extract_applicants(record):
    try:
        if isinstance(record, str):
            record = ast.literal_eval(record)
        applicants = record.get("parties", {}).get("applicants", [])
        names = [a.get("extracted_name", {}).get("value") for a in applicants]
        return [n for n in names if n]
    except:
        return []
    
final_df["applicant_list"] = final_df["biblio"].apply(extract_applicants)
```

```python
# 2. extract owners
def extract_owners(record):
    try:
        if isinstance(record, str):
            record = ast.literal_eval(record)
        owners = record.get("parties", {}).get("owners_all", [])
        names = [o.get("extracted_name", {}).get("value") for o in owners]
        return [n for n in names if n]
    except:
        return []

final_df["owner_list"] = final_df["biblio"].apply(extract_owners)
print(f"No owners: {final_df['owner_list'].apply(lambda x: len(x) == 0).sum()} / {len(final_df)}")

# 3. create assignee
final_df["assignee_list"] = final_df.apply(
    lambda row: row["owner_list"] if len(row["owner_list"]) > 0 else row["applicant_list"],
    axis=1
)
```
